```python
import matplotlib.pyplot as plt

# Data
Eg = [1.10, 1.91, 2.34, 2.64, 2.70]
Vg = [0.59, 1.78, 2.38, 2.62, 2.65]

labels = ['1N914', 'Red LED', 'Green LED', 'Blue LED', 'White LED']

plt.figure(figsize=(8,6))

plt.plot(Eg, Vg, 'o-', linewidth=2, markersize=8)

for i in range(len(labels)):
    plt.annotate(labels[i],
                 (Eg[i], Vg[i]),
                 textcoords="offset points",
                 xytext=(5,5))

plt.xlabel('Band Gap, $E_g$ (eV)')
plt.ylabel('Cut-in Voltage, $V_\\gamma$ (V)')
plt.title('Variation of Cut-in Voltage with Band Gap')
plt.grid(True)

plt.tight_layout()
plt.savefig('Vgamma_vs_Eg.png', dpi=300)
plt.show()
```
